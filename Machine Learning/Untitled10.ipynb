```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/data.csv')
df
```

```python
x=df[['Temperature']]
y=df['Pressure']
```

```python
from sklearn.linear_model import LinearRegression
lnr=LinearRegression()
lnr.fit(x,y)
y_pred=lnr.predict(x)
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.scatter(x,y_pred)
plt.plot(x,y_pred)
plt.show()
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
print('r2score',r2_score(y,y_pred),'\n','MAE',mean_absolute_error(y,y_pred),'\n','MSE',mean_squared_error(y,y_pred),'\n','RMSE',root_mean_squared_error(y,y_pred))
```

```python
data={'real':y,'predicted':y_pred}
res=pd.DataFrame(data)
res
```

```python
from sklearn.preprocessing import PolynomialFeatures
plf=PolynomialFeatures()
x_poly=plf.fit_transform(x)
```

```python
lnr.fit(x_poly,y)
y_plf=lnr.predict(x_poly)
```

```python
print('r2score',r2_score(y,y_plf),'\n','MAE',mean_absolute_error(y,y_plf),'\n','MSE',mean_squared_error(y,y_plf),'\n','RMSE',root_mean_squared_error(y,y_plf))
```

```python
plt.scatter(x,y)
plt.scatter(x,y_plf)
plt.plot(x,y_plf)
plt.show()
```
