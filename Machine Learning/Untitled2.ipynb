```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'/content/online_store_sales_modified.csv')
df
```

```python
df.isna().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mode()[0])
df['City']=df['City'].fillna(df['City'].mode()[0])
```

```python
df
```

```python
df['Payment_Method']=df['Payment_Method'].fillna(df['Payment_Method'].ffill)
```

```python
df=df.fillna(0)
```

```python
df.isna().sum()
```

```python
#Which age group spends the most money overall? - 26
df1=df.groupby('Age')['Price'].sum().sort_values().tail(1)
```

```python
df1
```

```python
#2. Do male or female customers purchase more on average? - F
dm=df.groupby('Gender')['Price'].mean()
dm
```

```python
#3. Which city contributes the highest revenue? - Delhi
DC=df.groupby('City')['Price'].sum().sort_values().tail(1)
DC
```

```python
#4. Who are the top 5 customers by total spending? - sophia,chloe,aisha,grace,liam
dn=df.groupby('Customer_Name')['Price'].sum().sort_values(ascending=False).head()
dn
```

```python
#5. Which product category has the highest sales volume? - Books
dcc=df.groupby('Product_Category')['Quantity'].sum().sort_values().tail(1)
dcc
```

```python
#6. What are the top 3 most sold products? - comics , milk , jacket
dp=df.groupby('Product')['Quantity'].sum().sort_values(ascending=False).head(3)
dp
```

```python
#What is the average discount given per product category?
dd=df.groupby('Product_Category')['Discount (%)'].mean()
dd
```
