```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/DT_data.csv')
df
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df['a1']=encoder.fit_transform(df['a1'])
df['a2']=encoder.fit_transform(df['a2'])
df['a3']=encoder.fit_transform(df['a3'])
df
```

```python
x=df.drop('Outcome',axis=1)
y=df['Outcome']
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc=DecisionTreeClassifier(criterion='entropy')
dtc.fit(x,y)
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))
tree.plot_tree(dtc,feature_names=['a1','a2','a3'],class_names=['no','yes'],filled=True)
plt.show()
```
