```python
import pandas as pd
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/Iris.csv')
df
```

```python
df.drop('Id',axis=1,inplace=True)
df
```

```python
x=df.drop('Species',axis=1)
y=df['Species']
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc=DecisionTreeClassifier(criterion='entropy')
dtc.fit(x,y)
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
plt.figure(figsize=(30,30))
tree.plot_tree(dtc,feature_names=['sepal l','sepal w','petal l','petal w'],class_names=['setosa','versicolor','virginica'])
plt.show()
```
