```python
import numpy as np
import pandas as pd
df=pd.read_csv(r'/content/diabetes.csv')
```

```python
x=df.drop(columns='Outcome')
y=df['Outcome']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
x_tr,x_te,y_tr,y_te=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
sc=StandardScaler()
x_tr_sc=sc.fit_transform(x_tr)
x_te_sc=sc.transform(x_te)
```

```python
knn=KNeighborsClassifier(n_neighbors=9)
knn.fit(x_tr_sc,y_tr)
pred=knn.predict(x_te_sc)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
print(classification_report(pred,y_te))
```

```python
knn1=KNeighborsClassifier()
knn1.fit(x_tr,y_tr)
pred1=knn1.predict(x_te)
print(classification_report(pred1,y_te))
```

```python
 from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
```

```python
gnb.fit(x_tr_sc,y_tr)
predgnb=gnb.predict(x_te_sc)
```

```python
print(classification_report(y_te,predgnb))
```
