```python
import numpy as np
import pandas as pd
df=pd.read_csv(r'/content/diabetes.csv')
```

```python
df.shape
```

```python
df.dtypes
```

```python
y=df['Outcome']
y.ndim
```

```python
x=df.drop('Outcome',axis=1)
x.ndim
```

```python
from sklearn.model_selection import train_test_split
x_tr,x_te,y_tr,y_te=train_test_split(x,y,test_size=0.30)
```

```python
x_tr.shape
```

```python
x_te.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_tr)
x_tr=scaler.transform(x_tr)
x_te=scaler.transform(x_te)
x_tr
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=8)
knn.fit(x_tr,y_tr)
y_pred=knn.predict(x_te)
```

```python
y_pred
```

```python
y_te.values
```

```python
print(knn.predict(scaler.transform([[6,148,72,35,0,33.6,0.627,50]])))
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
def cscore(test,predd):
  print(accuracy_score(test,predd))
  print(confusion_matrix(test,predd))
  print(classification_report(test,predd))
```

```python
cscore(y_te,y_pred)
```
