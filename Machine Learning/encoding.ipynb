```python
import pandas as pd
import numpy as np
```

```python
data={'colour':['red','green','blue','red','green'],
      'size':['small','medium','large','small','medium'],
      'material':['wood','wood','metal','wood','wood']}
```

```python
df=pd.DataFrame(data)
df
```

```python
df.columns=['Colour','Size','Material']
```

```python
df
```

```python
df1=df.copy()
```

```python
df1['Colour']=df1['Colour'].map({'red':0,'green':1,'blue':2})
df1['Material']=df1['Material'].map({'wood':0,'metal':1})
df1['Size']=df1['Size'].map({'small':0,'medium':1,'large':2})
df1
```

```python
df2=df.copy()
```

```python
from sklearn.preprocessing import LabelEncoder
encode_c=LabelEncoder()
encode_m=LabelEncoder()
encode_s=LabelEncoder()

df2['Colour']=encode_c.fit_transform(df2['Colour'])
df2['Material']=encode_m.fit_transform(df2['Material'])
df2['Size']=encode_s.fit_transform(df2['Size'])
df2
```

```python
df2['Colour']=encode_c.inverse_transform(df2['Colour'])
df2['Material']=encode_m.inverse_transform(df2['Material'])
df2['Size']=encode_s.inverse_transform(df2['Size'])
df2
```

```python
df3=df.copy()
cat_col=['Colour','Size','Material']
encoders=[LabelEncoder() for i in cat_col]
```

```python
for i,col in enumerate(cat_col):
  df3[col]=encoders[i].fit_transform(df3[col])
df3
```

```python
for i,col in enumerate(cat_col):
  df3[col]=encoders[i].inverse_transform(df3[col])
df3
```
