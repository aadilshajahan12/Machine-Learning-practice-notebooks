```python
import numpy as np
import pandas as pd
df=pd.read_csv(r'/content/glass_knn_numeric.csv')
```

```python
df
```

```python
df.isna().sum()
```

```python
x=df.columns
for i in x:
  print(i,df[i].unique())
```

```python
df['Na']=df['Na'].replace('fifteen',15)
df['Mg']=df['Mg'].replace('missing',np.nan)
df['Al']=df['Al'].replace('??',np.nan)
df['K']=df['K'].replace('none',np.nan)
df['Ba']=df['Ba'].replace('low',np.nan)
df['Fe']=df['Fe'].replace('zero',np.nan)
```

```python
for i in x:
  df[i]=df[i].astype(float)
  df[i]=df[i].fillna(df[i].mean())
```

```python
df.isna().sum()
```

```python
df['Al']
```

```python
x=df.drop('Type',axis=1)
x.ndim
y=df['Type']
y.ndim
```

```python
from sklearn.model_selection import train_test_split
x_tr,x_te,y_tr,y_te=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
sc.fit(x_tr)
x_tr=sc.transform(x_tr)
x_te=sc.transform(x_te)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(x_tr,y_tr)
y_pred=knn.predict(x_te)
```
