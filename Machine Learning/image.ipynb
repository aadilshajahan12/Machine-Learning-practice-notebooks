```python
from skimage.io import imread
```

```python
import os
path=os.listdir('/content/drive/MyDrive/Data')
for sub in path:
  subpath=os.path.join('/content/drive/MyDrive/Data',sub)
  images=os.listdir(subpath)
  for img in images:
    img_path=os.path.join(subpath,img)
    img_array=imread(img_path)
    print(img_array)
```
