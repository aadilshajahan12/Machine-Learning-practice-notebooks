```python
import pandas as pd
import numpy as np
d1={'id':[101,102,103,104,105],
    'fname':['eran','isgar','falton','yonah','esha'],
    'lname':['pling','hoon','kiljah','toorbani','kling'],
    'age':[22,34,33,16,29]}
d2={'prof':['eko','dr','engg','pipe','coder'],
    'id':[103,104,105,106,107],
    'loc':['sonpur','cairo','afg','egypt','morocco'],
    'sal':[20000,12000,44000,28000,60000]}
d1
```

```python
d2
```

```python
df1=pd.DataFrame(d1)
df1
```

```python
df2=pd.DataFrame(d2)
df2
```

```python
df3=pd.merge(df1,df2,on='id',how='inner')[['fname','lname']]
df3
```

```python
df4=pd.merge(df1,df2,how='left')
df4
```

```python
df4=pd.merge(df1,df2,how='right')
df4
```

```python
df4=pd.merge(df1,df2,how='outer')
df4
```
