```python
import matplotlib.pyplot as plt
x=[1,2,4,3]
y=[5,6,3,9]
plt.scatter(x,y,color='red')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('EXAMPLE')
plt.plot(x,y)
plt.show()
```

```python
plt.plot(y,'*:b')
plt.show()
```

```python
plt.plot(x,y,'*:r',markerfacecolor='blue',ms=11)
```

```python
x1=[1,3,2,6,9,3]
y1=[6,1,7,4,7,8]
plt.subplot(2,2,2)
plt.plot(x1,y1)
plt.subplot(2,2,1)
plt.plot(x,y)
plt.suptitle('ANANDU')
plt.show()
```

```python
year=[2021,2022,2023,2024]
points=[85,88,83,89]
goals=[76,88,93,81]
wins=[22,24,20,25]
plt.subplot(3,2,1)
plt.plot(year,points)
plt.subplot(3,2,2)
plt.plot(year,goals)
plt.subplot(3,2,3)
plt.plot(year,wins)
plt.subplots_adjust(hspace=1.5)

plt.subplot(3,2,4)
plt.plot(goals,wins)
plt.show()
```

```python
w=[88,56,68,95,78]
h=[168,159,175,190,188]
plt.plot(w,h,marker='o')
plt.xlabel('Weight')
plt.ylabel('Height')
plt.title('W vs H')
plt.grid(linestyle='--')
plt.show()
```

```python
plt.bar(w,h,width=6,color='red')
```

```python
plt.barh(w,h,color='k')
```
