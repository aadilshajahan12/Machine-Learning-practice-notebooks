```python
import pandas as pd
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/poly_Ice_cream selling data.csv')
df
```

```python
x=df.iloc[:,:-1]
y=df['Ice Cream Sales (units)']
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
plf=PolynomialFeatures()
x_poly=plf.fit_transform(x)
```

```python
from sklearn.linear_model import LinearRegression
lnr=LinearRegression()
lnr.fit(x_poly,y)
y_pred=lnr.predict(x_poly)
```

```python
plt.scatter(x,y)
plt.scatter(x,y_pred)
plt.plot(x,y_pred,color='red')
plt.show()
```
