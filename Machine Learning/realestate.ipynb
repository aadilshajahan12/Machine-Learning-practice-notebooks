```python
import pandas as pd
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/Real estate.csv')
df
```

```python
df.drop(columns=['No','X1 transaction date'],inplace=True)
x=df.iloc[:,:-1]
y=df['Y house price of unit area']
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_scale=scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
x_tr,x_te,y_tr,y_te=train_test_split(x_scale,y,test_size=0.30,random_state=11)
```

```python
from sklearn.linear_model import LinearRegression
lnr=LinearRegression()
lnr.fit(x_tr,y_tr)
y_lnr=lnr.predict(x_te)
```

```python
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor()
knn.fit(x_tr,y_tr)
y_knn=knn.predict(x_te)
```

```python
from sklearn.metrics import r2_score
print("lnr:",r2_score(y_te,y_lnr))
print("knn:",r2_score(y_te,y_knn))
```
