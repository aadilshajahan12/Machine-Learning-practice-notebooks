```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/Salary_Data.csv')
df
```

```python
x=df.iloc[:,:-1].values
y=df['Salary']
x.ndim
```

```python
from sklearn.model_selection import train_test_split
x_tr,x_te,y_tr,y_te=train_test_split(x,y,test_size=0.2,random_state=11)
```

```python
from sklearn.linear_model import LinearRegression
lnr=LinearRegression()
lnr.fit(x_tr,y_tr)
y_pred=lnr.predict(x_te)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
r2_score(y_te,y_pred)
```

```python
print(mean_absolute_error(y_te,y_pred), '\n' ,mean_squared_error(y_te,y_pred), '\n', root_mean_squared_error(y_te,y_pred))
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_te,y_pred)
plt.scatter(x_te,y_te)
plt.plot(x_te,y_pred)
plt.xlabel("years")
plt.ylabel('Salary')
plt.show()
```
