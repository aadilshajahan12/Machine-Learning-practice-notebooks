```python
import pandas as pd
df_te=pd.read_csv(r'/content/drive/MyDrive/Datasets/SMS_test.csv',encoding='ISO-8859-1')
x_te=df_te.drop(columns=['S. No.','Label'])
y_te=df_te['Label']
```

```python
df_tr=pd.read_csv(r'/content/drive/MyDrive/Datasets/SMS_train.csv',encoding='ISO-8859-1')
x_tr=df_tr.drop(columns=['S. No.','Label'])
y_tr=df_tr['Label']
```

```python
y_te=y_te.map({'Spam':1,'Non-Spam':0})
```

```python
y_tr=y_tr.map({'Spam':1,'Non-Spam':0})
y_tr
```

```python
x_tr
```

```python
x_tr=x_tr['Message_body']
x_tr=x_tr.str.replace('[^A-z ]','',regex=True)
x_te=x_te['Message_body']
x_te=x_te.str.replace('[^A-z ]','',regex=True)
```

```python
x_tr
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
w=stopwords.words('english')
from nltk.stem import SnowballStemmer
nltk.download('punkt_tab')
sbs=SnowballStemmer('english')
x_tr=x_tr.apply(lambda line:[sbs.stem(word) for word in nltk.word_tokenize(line) if len(word)>1 and word.lower() not in w]).apply(lambda s:' '.join(s))
x_te=x_te.apply(lambda line:[sbs.stem(word) for word in nltk.word_tokenize(line) if len(word)>1 and word.lower() not in w]).apply(lambda s:' '.join(s))
```

```python
x_tr
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vect=TfidfVectorizer()
tr=vect.fit_transform(x_tr)
te=vect.transform(x_te)
```

```python
tr
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB,BernoulliNB
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier
from sklearn.metrics import classification_report
models=[KNeighborsClassifier(),GaussianNB(),BernoulliNB(),RandomForestClassifier(random_state=11),AdaBoostClassifier(random_state=11)]
for obj in models:
  obj.fit(tr.toarray(),y_tr)
  y_pred=obj.predict(te.toarray())
  print(obj)
  print(classification_report(y_te,y_pred))
```
