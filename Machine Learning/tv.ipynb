```python
import pandas as pd
import numpy as np
df=pd.read_csv(r'/content/drive/MyDrive/Datasets/tvmarketing.csv')
df
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
x=df.iloc[:,:-1]
y=df['Sales']
```

```python
from sklearn.model_selection import train_test_split
x_tr,x_te,y_tr,y_te=train_test_split(x,y,test_size=0.30,random_state=11)
```

```python
from sklearn.linear_model import LinearRegression
lnr=LinearRegression()
lnr.fit(x_tr,y_tr)
```

```python
lnr.coef_*222.4+lnr.intercept_
```

```python
y_pred=lnr.predict(x_te)
y_pred
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
print('r2 score',r2_score(y_te,y_pred))
print('mae',mean_absolute_error(y_te,y_pred))
print('mse',mean_squared_error(y_te,y_pred))
print('rmse',root_mean_squared_error(y_te,y_pred))
```

```python
from matplotlib.typing import ColorType
import matplotlib.pyplot as plt
plt.scatter(x_tr,y_tr)
plt.scatter(x_te,y_pred)
plt.xlabel('TV (in thousands)')
plt.ylabel('Sales (in millions)')
plt.show()
```
